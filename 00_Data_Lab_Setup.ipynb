## 3. Quick smoke test: load data and draw a plot

```python
import matplotlib.pyplot as plt
df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")
print("Loaded Titanic dataset:", df.shape)
df.head()
```

```python
sns.set_theme(style="whitegrid")
ax = sns.countplot(data=df, x="Pclass", color="#2a6fdb")
ax.set(title="Passengers by class (setup test plot)", xlabel="Passenger class", ylabel="Count")
plt.show()
```
